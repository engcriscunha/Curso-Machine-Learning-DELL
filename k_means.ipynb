#### K-MEANS

##### Importações

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

##### Leitura dos dados

```python
df = pd.read_csv('Mall_Customers.csv')
df.head()
```

```python
df.shape
```

##### Verificar dados nulos

```python
df.isnull().sum()
```

##### Informações estatísticas

```python
df.describe()
```

##### Gráfico renda anual x score

```python
plt.scatter(df['Annual Income (k$)'], df['Spending Score (1-100)'], marker='.', color='blue')
plt.xlabel('Renda Anual (k$)')
plt.ylabel('Spending Score (1-100)')    
plt.title('Annual Income vs Spending Score')
plt.show()
```

##### Selecionando dados para agrupamento

```python
x = df[['Annual Income (k$)', 'Spending Score (1-100)']]
x.head()
```

##### Importando K-Means

```python
from sklearn.cluster import KMeans
```

##### Clusterização com K =5 

```python
modelo_kmeans = KMeans(n_clusters=5, init='k-means++')
```

```python
y_kmeans = modelo_kmeans.fit_predict(x)
```

```python
print(y_kmeans)
```

##### Vizualizando o primeiro grupo

```python
print(x[y_kmeans == 0].head())
```

##### Vizualizando os grupos

```python
k_grupos = 5
cores = ['red', 'blue', 'green', 'cyan', 'magenta']
for k in range(k_grupos):
    cluster = x[y_kmeans == k]
    plt.scatter(cluster['Annual Income (k$)'], cluster['Spending Score (1-100)'], s=100, c=cores[k], label=f'Cluster {k}')


plt.title('Grupos de Clientes')
plt.xlabel('Renda Anual (k$)')  
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.grid()
plt.show()
```
