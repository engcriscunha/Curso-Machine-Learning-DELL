```python
frutas['fruit_name'] == 'apple' #aparece True para as linhas em que a fruta é maçã e False para as demais
```

```python
macas = frutas['fruit_name'] == 'apple' #aparece as linhas em que a fruta é maçã
```

```python
frutas[macas]
```

```python
frutas['mass']>175
```

```python
frutas[macas & frutas['mass']>175] #aparece as linhas em que a fruta é maçã e o peso é maior que 175
```

```python
pesadas = frutas['mass']>175 #aparece as linhas em que o peso é maior que 175
```

```python
frutas[macas & pesadas] #aparece as linhas em que a fruta é maçã e o peso é maior que 175
```
