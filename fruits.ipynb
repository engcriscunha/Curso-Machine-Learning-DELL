#### ENCONTRANDO OUTLIERS

```python
data = pd.read_table('fruit_data_with_colors_miss.txt', na_values=['.','?'])
data.fillna(data.mode().iloc[0], inplace=True)
```

```python
macas = data[data['fruit_name']== 'apple']
```

```python
macas.describe()
```

```python
estatistica = macas['mass'].describe()
```

```python
#OutLiers que são maiores que a média e duas vezes o desvio padrão


macas[(macas['mass'] > estatistica['mean'] + (estatistica['std'])*2)]
```

```python
#OutLiers que são menores que a média e duas vezes o desvio padrão

macas[(macas['mass'] < estatistica['mean'] - (estatistica['std'])*2)]
```

```python
estatistica['mean'] #média
```

```python
estatistica['std'] #desvio padrão
```
